```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/water_potability_cleaned.csv")

print("Dataset shape:", df.shape)
print(df.head())
```

```python
print("\nStatistical Summary:")
print(df.describe())
```

```python
plt.figure(figsize=(6, 4))

sns.countplot(x="Potability", data=df)

plt.title("Water Potability Distribution")
plt.xlabel("Potability")
plt.ylabel("Number of Samples")

plt.show()
```

```python
df.hist(
    figsize=(15, 12),
    bins=30
)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15, 8))

sns.boxplot(data=df)

plt.xticks(rotation=45)
plt.title("Box Plot of Water Quality Features")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="ph",
    kde=True
)

plt.title("Distribution of Water pH")
plt.xlabel("pH")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Potability",
    y="ph"
)

plt.title("pH vs Water Potability")
plt.xlabel("Potability")
plt.ylabel("pH")

plt.show()
```

```python
features = [
    "ph",
    "Hardness",
    "Solids",
    "Chloramines",
    "Sulfate",
    "Conductivity",
    "Organic_carbon",
    "Trihalomethanes",
    "Turbidity"
]

for feature in features:

    plt.figure(figsize=(7, 4))

    sns.boxplot(
        data=df,
        x="Potability",
        y=feature
    )

    plt.title(f"{feature} vs Potability")
    plt.xlabel("Potability")
    plt.ylabel(feature)

    plt.show()
```

```python
plt.savefig("../plots/potability_distribution.png")
```
